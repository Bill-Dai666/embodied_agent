```python
import pandas as pd

# 读取数据文件
db_avatar = pd.read_csv('db_avatar.csv')
db_text = pd.read_csv('db_text.csv')
qualtrics = pd.read_csv('qualtrics_raw_data.csv')

# 创建 ResponseID -> article_id 的映射
response_to_article = dict(zip(qualtrics['ResponseID'], qualtrics['article_id']))

print(f'db_avatar 行数: {len(db_avatar)}')
print(f'db_text 行数: {len(db_text)}')
print(f'qualtrics ResponseID 数量: {len(response_to_article)}')

# 处理函数：将一行数据拆分成两条记录（user_input 和 bot_response）
def process_row(row, source_type):
    """处理一行数据，返回两条记录（user_input 和 bot_response）"""
    records = []
    response_id = row['ResponseID']
    article_id = response_to_article.get(response_id, None)
    turn_number = row['turn_number']
    
    # 确定 person_id_2
    if source_type == 'avatar':
        person_id_2 = 'avatar_based_chatbot'
        bot_speaker_id = 'avatar'
    else:  # text
        person_id_2 = 'text_based_chatbot'
        bot_speaker_id = 'text'
    
    # 处理 user_input（如果存在且不是初始提示）
    if pd.notna(row['user_input']) and row['user_input'] != '[initial prompt trigger]':
        records.append({
            'person_id': response_id,
            'article_id': article_id,
            'turn_number': turn_number,
            'speaker_id': response_id,
            'text': row['user_input'],
            'person_id_1': response_id,
            'person_id_2': person_id_2,
            'source_type': source_type,
            'is_user': True
        })
    
    # 处理 bot_response（如果存在）
    if pd.notna(row['bot_response']):
        records.append({
            'person_id': response_id,
            'article_id': article_id,
            'turn_number': turn_number,
            'speaker_id': bot_speaker_id,
            'text': row['bot_response'],
            'person_id_1': response_id,
            'person_id_2': person_id_2,
            'source_type': source_type,
            'is_user': False
        })
    
    return records

# 处理 db_avatar
avatar_records = []
for _, row in db_avatar.iterrows():
    avatar_records.extend(process_row(row, 'avatar'))

# 处理 db_text
text_records = []
for _, row in db_text.iterrows():
    text_records.extend(process_row(row, 'text'))

# 合并所有记录
all_records = avatar_records + text_records
df_all = pd.DataFrame(all_records)

print(f'\n合并后总记录数: {len(df_all)}')

# 按 ResponseID 分组，分配 conversation_id
df_all = df_all.sort_values(['person_id', 'article_id', 'turn_number', 'is_user'])

# 分配 conversation_id（从1开始，每个 ResponseID 一个 conversation_id）
unique_person_ids = df_all['person_id'].unique()
person_to_conversation = {pid: idx + 1 for idx, pid in enumerate(unique_person_ids)}
df_all['conversation_id'] = df_all['person_id'].map(person_to_conversation)

# 在每个 conversation_id 内分配 turn_id（从1开始）
# 排序：先按 conversation_id，再按 turn_number，最后按 is_user（True在前，即user_input在前）
df_all = df_all.sort_values(['conversation_id', 'turn_number', 'is_user'], ascending=[True, True, False])

turn_ids = []
current_conversation = None
current_turn = 0

for _, row in df_all.iterrows():
    if row['conversation_id'] != current_conversation:
        current_conversation = row['conversation_id']
        current_turn = 0
    
    current_turn += 1
    turn_ids.append(current_turn)

df_all['turn_id'] = turn_ids

# 选择最终需要的列（与 simulated_conversations.csv 一致）
final_df = df_all[['person_id', 'article_id', 'conversation_id', 'turn_id', 'speaker_id', 'text', 'person_id_1', 'person_id_2']].copy()

# 保存为 conversation.csv
final_df.to_csv('conversation.csv', index=False, encoding='utf-8')

print(f'\n✓ 已生成 conversation.csv')
print(f'最终记录数: {len(final_df)}')
print(f'唯一 conversation_id 数: {final_df["conversation_id"].nunique()}')
print(f'\n前5行预览:')
print(final_df.head())
```

```python
import pandas as pd

# 读取两个文件
qualtrics_df = pd.read_csv('qualtrics_raw_data.csv')
prolific_df = pd.read_csv('prolific_raw_data.csv')

print(f'qualtrics_raw_data.csv 行数: {len(qualtrics_df)}')
print(f'prolific_raw_data.csv 行数: {len(prolific_df)}')

# 选择 prolific 中需要的列：Participant id, Age, Sex, Ethnicity simplified
prolific_selected = prolific_df[['Participant id', 'Age', 'Sex', 'Ethnicity simplified']].copy()

# 重命名 Participant id 为 prolific_id 以便合并
prolific_selected = prolific_selected.rename(columns={'Participant id': 'prolific_id'})

print(f'\nprolific 中选中的列: {list(prolific_selected.columns)}')

# 通过 prolific_id 进行左连接（保留 qualtrics 中的所有记录）
merged_df = qualtrics_df.merge(
    prolific_selected,
    on='prolific_id',
    how='left'
)

print(f'\n合并后行数: {len(merged_df)}')
print(f'合并后列数: {len(merged_df.columns)}')

# 检查匹配情况
matched_count = merged_df['Age'].notna().sum()
print(f'\n成功匹配的记录数: {matched_count}')
print(f'未匹配的记录数: {len(merged_df) - matched_count}')

# 保存为新文件 survey_responses_first.csv（不修改源文件）
merged_df.to_csv('survey_responses_first.csv', index=False, encoding='utf-8')

print(f'\n✓ 已生成 survey_responses_first.csv')
print(f'\n前3行预览（显示新增的列）:')
print(merged_df[['ResponseID', 'prolific_id', 'Age', 'Sex', 'Ethnicity simplified']].head(3))
```

```python
import pandas as pd

# 读取文件
survey_first = pd.read_csv('survey_responses_first.csv')
conversation = pd.read_csv('conversation.csv')

print(f'survey_responses_first.csv 行数: {len(survey_first)}')
print(f'conversation.csv 行数: {len(conversation)}')

# 从 conversation.csv 中获取每个 ResponseID 对应的 condition
# 通过 speaker_id 判断：如果是 "avatar" 就是 "avatar"，如果是 "text" 就是 "text"
# 只取 bot 的 speaker_id（不是 ResponseID 本身的）
conversation_bot = conversation[conversation['speaker_id'].isin(['avatar', 'text'])].copy()
response_to_condition = {}
for response_id in conversation_bot['person_id'].unique():
    # 获取该 ResponseID 的第一个 bot speaker_id
    bot_rows = conversation_bot[conversation_bot['person_id'] == response_id]
    if len(bot_rows) > 0:
        speaker = bot_rows.iloc[0]['speaker_id']
        response_to_condition[response_id] = speaker

print(f'\n找到 condition 的 ResponseID 数量: {len(response_to_condition)}')

# 过滤掉表头行（如果存在）
survey_first = survey_first[survey_first['ResponseID'].notna()].copy()
survey_first = survey_first[~survey_first['ResponseID'].str.startswith('ResponseID', na=False)].copy()
survey_first = survey_first[~survey_first['ResponseID'].str.startswith('{', na=False)].copy()

print(f'\n过滤后 survey_responses_first 行数: {len(survey_first)}')

# 定义 P 字段映射（P1_1 到 P4_5 对应 P1 到 P20）
p_mapping = {
    'P1_1': 'P1_have_vivid_imagination',
    'P1_2': 'P2_hold_a_grudge',
    'P1_3': 'P3_do_not_mind_centre_of_attention',
    'P1_4': 'P4_do_not_like_poetry',
    'P1_5': 'P5_complete_tasks_successfully',
    'P2_1': 'P6_believe_others_have_good_intentions',
    'P2_2': 'P7_avoid_philosophical_discussions',
    'P2_3': 'P8_need_a_push_to_get_started',
    'P2_4': 'P9_cut_others_to_pieces',
    'P2_5': 'P10_make_friends_easily',
    'P3_1': 'P11_feel_comfortable_with_myself',
    'P3_2': 'P12_often_feel_blue',
    'P3_3': 'P13_am_easy_to_satisfy',
    'P3_4': 'P14_keep_in_the_background',
    'P3_5': 'P15_am_always_prepared',
    'P4_1': 'P16_enjoy_wild_flights_of_fantasy',
    'P4_2': 'P17_get_stressed_out_easily',
    'P4_3': 'P18_avoid_contact_with_others',
    'P4_4': 'P19_am_not_easily_bothered',
    'P4_5': 'P20_shirk_my_duties'
}

# 定义 Q 字段映射（Q1 到 Q29_3 对应 PCQ_1 到 G_SAFETY_3）
q_mapping = {
    # PCQ_1 到 PCQ_14 (Q1 到 Q14)
    'Q1': 'PCQ_1', 'Q2': 'PCQ_2', 'Q3': 'PCQ_3', 'Q4': 'PCQ_4', 'Q5': 'PCQ_5',
    'Q6': 'PCQ_6', 'Q7': 'PCQ_7', 'Q8': 'PCQ_8', 'Q9': 'PCQ_9', 'Q10': 'PCQ_10',
    'Q11': 'PCQ_11', 'Q12': 'PCQ_12', 'Q13': 'PCQ_13', 'Q14': 'PCQ_14',
    # EMP_1 到 EMP_5 (Q15 到 Q19)
    'Q15': 'EMP_1', 'Q16': 'EMP_2', 'Q17': 'EMP_3', 'Q18': 'EMP_4', 'Q19': 'EMP_5',
    # TRUST_1 到 TRUST_5 (Q20 到 Q24)
    'Q20': 'TRUST_1', 'Q21': 'TRUST_2', 'Q22': 'TRUST_3', 'Q23': 'TRUST_4', 'Q24': 'TRUST_5',
    # G_ANTHRO_1 到 G_ANTHRO_5 (Q25_1 到 Q25_5)
    'Q25_1': 'G_ANTHRO_1', 'Q25_2': 'G_ANTHRO_2', 'Q25_3': 'G_ANTHRO_3', 
    'Q25_4': 'G_ANTHRO_4', 'Q25_5': 'G_ANTHRO_5',
    # G_ANIMACY_1 到 G_ANIMACY_6 (Q26_1 到 Q26_6)
    'Q26_1': 'G_ANIMACY_1', 'Q26_2': 'G_ANIMACY_2', 'Q26_3': 'G_ANIMACY_3',
    'Q26_4': 'G_ANIMACY_4', 'Q26_5': 'G_ANIMACY_5', 'Q26_6': 'G_ANIMACY_6',
    # G_LIKE_1 到 G_LIKE_5 (Q27_1 到 Q27_5)
    'Q27_1': 'G_LIKE_1', 'Q27_2': 'G_LIKE_2', 'Q27_3': 'G_LIKE_3',
    'Q27_4': 'G_LIKE_4', 'Q27_5': 'G_LIKE_5',
    # G_INTEL_1 到 G_INTEL_5 (Q28_1 到 Q28_5)
    'Q28_1': 'G_INTEL_1', 'Q28_2': 'G_INTEL_2', 'Q28_3': 'G_INTEL_3',
    'Q28_4': 'G_INTEL_4', 'Q28_5': 'G_INTEL_5',
    # G_SAFETY_1 到 G_SAFETY_3 (Q29_1 到 Q29_3)
    'Q29_1': 'G_SAFETY_1', 'Q29_2': 'G_SAFETY_2', 'Q29_3': 'G_SAFETY_3'
}

# 创建新的数据框
new_rows = []

for _, row in survey_first.iterrows():
    response_id = row['ResponseID']
    
    # 获取 condition
    condition = response_to_condition.get(response_id, '')
    
    # 创建新行
    new_row = {
        'person_id': response_id,
        'archetype': '',
        'condition': condition,
        'article_id': row['article_id'] if pd.notna(row['article_id']) else ''
    }
    
    # 添加 P 字段（P1_1 到 P4_5）
    for old_col, new_col in p_mapping.items():
        new_row[new_col] = row[old_col] if old_col in row and pd.notna(row[old_col]) else ''
    
    # 添加 Q 字段（Q1 到 Q29_3）
    for old_col, new_col in q_mapping.items():
        new_row[new_col] = row[old_col] if old_col in row and pd.notna(row[old_col]) else ''
    
    # 添加 demographic 字段
    new_row['Age'] = row['Age'] if pd.notna(row['Age']) else ''
    new_row['Sex'] = row['Sex'] if pd.notna(row['Sex']) else ''
    new_row['Ethnicity simplified'] = row['Ethnicity simplified'] if pd.notna(row['Ethnicity simplified']) else ''
    
    new_rows.append(new_row)

# 创建 DataFrame
new_df = pd.DataFrame(new_rows)

# 按照 simulated_survey_responses.csv 的列顺序排列
column_order = [
    'person_id', 'archetype', 'condition', 'article_id',
    'P1_have_vivid_imagination', 'P2_hold_a_grudge', 'P3_do_not_mind_centre_of_attention',
    'P4_do_not_like_poetry', 'P5_complete_tasks_successfully', 'P6_believe_others_have_good_intentions',
    'P7_avoid_philosophical_discussions', 'P8_need_a_push_to_get_started', 'P9_cut_others_to_pieces',
    'P10_make_friends_easily', 'P11_feel_comfortable_with_myself', 'P12_often_feel_blue',
    'P13_am_easy_to_satisfy', 'P14_keep_in_the_background', 'P15_am_always_prepared',
    'P16_enjoy_wild_flights_of_fantasy', 'P17_get_stressed_out_easily', 'P18_avoid_contact_with_others',
    'P19_am_not_easily_bothered', 'P20_shirk_my_duties',
    'PCQ_1', 'PCQ_2', 'PCQ_3', 'PCQ_4', 'PCQ_5', 'PCQ_6', 'PCQ_7', 'PCQ_8', 'PCQ_9', 'PCQ_10',
    'PCQ_11', 'PCQ_12', 'PCQ_13', 'PCQ_14',
    'EMP_1', 'EMP_2', 'EMP_3', 'EMP_4', 'EMP_5',
    'TRUST_1', 'TRUST_2', 'TRUST_3', 'TRUST_4', 'TRUST_5',
    'G_ANTHRO_1', 'G_ANTHRO_2', 'G_ANTHRO_3', 'G_ANTHRO_4', 'G_ANTHRO_5',
    'G_ANIMACY_1', 'G_ANIMACY_2', 'G_ANIMACY_3', 'G_ANIMACY_4', 'G_ANIMACY_5', 'G_ANIMACY_6',
    'G_LIKE_1', 'G_LIKE_2', 'G_LIKE_3', 'G_LIKE_4', 'G_LIKE_5',
    'G_INTEL_1', 'G_INTEL_2', 'G_INTEL_3', 'G_INTEL_4', 'G_INTEL_5',
    'G_SAFETY_1', 'G_SAFETY_2', 'G_SAFETY_3',
    'Age', 'Sex', 'Ethnicity simplified'
]

# 确保所有列都存在
for col in column_order:
    if col not in new_df.columns:
        new_df[col] = ''

# 按顺序选择列
final_df = new_df[column_order].copy()

# 保存为新文件 survey_responses.csv
final_df.to_csv('survey_responses.csv', index=False, encoding='utf-8')

print(f'\n✓ 已生成 survey_responses.csv')
print(f'最终记录数: {len(final_df)}')
print(f'最终列数: {len(final_df.columns)}')
print(f'\n前3行预览:')
print(final_df.head(3))
```
